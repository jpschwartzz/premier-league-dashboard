```python
import pandas as pd
import numpy as np 
import plotly.express as px
pd.set_option('display.max_columns', None)

caminho_players = '../data/processed/df_dados_players.csv'
caminho_teams = '../data/processed/df_dados_teams.csv'

df_players = pd.read_csv(caminho_players, index_col=0)
df_teams = pd.read_csv(caminho_teams)

# Criando variável artilheiros
df_artilheiros = df_players.sort_values(by='Gls', ascending=False).head(10)

# Criando gráfico artilheiros
grafico_artilheiros = px.bar(df_artilheiros, x='Player', y='Gls', labels={'Player': 'Jogador', 'Gls': 'Gols'})
grafico_artilheiros.update_layout(title='Top 10 artilheiros da Premier League 2024/25', title_x=0.5, xaxis_tickangle=-45)

# Criando variável assistentes
df_assistentes = df_players.sort_values(by='Ast', ascending=False).head(10)

# Criando gráfico assistentes
grafico_assistentes = px.bar(df_assistentes, x='Player', y='Ast', labels={'Player': 'Jogador', 'Ast': 'Assistências'})
grafico_assistentes.update_layout(title='Top 10 assistentes da Premier League 2024/25', title_x=0.5, xaxis_tickangle=-45)

# Criando variável decisivos
df_decisivos = df_players.sort_values(by='G+A', ascending=False).head(10)

# Mudando o nome das colunas `Gls` e `Ast`
df_decisivos = df_decisivos.rename(columns={'Gls': 'Gols', 'Ast': 'Assistências'})

# Criando gráfico decisivos
grafico_decisivos = px.bar(df_decisivos, x='Player', y=['Gols', 'Assistências'], labels={'Player': 'Jogador', 'G+A': 'Gols + Assistências', 'value': 'Gols + Assistências'})
grafico_decisivos.update_layout(title='Top 10 jogadores mais decisivos da Premier League 2024/25', title_x=0.5, xaxis_tickangle=-45, legend_title_text='')

 
```

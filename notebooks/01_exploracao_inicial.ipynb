```python
import pandas as pd
pd.set_option('display.max_columns', None)

caminho_dados_matches = '../data/raw/PL_matches_2000_2025.csv'
caminho_dados_players = '../data/raw/PL_players_2024_25.csv'
caminho_dados_teams = '../data/raw/PL_teams_2024_25.csv'

df_dados_matches = pd.read_csv(caminho_dados_matches)
df_dados_players = pd.read_csv(caminho_dados_players)
df_dados_teams = pd.read_csv(caminho_dados_teams)

# Renomeando colunas com nomes iguais para a mesma estatística só que a cada 90 minutos.
df_dados_players.rename(columns={
    'Gls.1': 'Gls_per90',
    'Ast.1': 'Ast_per90',
    'G+A.1': 'G+A_per90',
    'G-PK.1': 'G-PK_per90',
    'xG.1': 'xG_per90',
    'xAG.1': 'xAG_per90',
    'npxG.1': 'npxG_per90',
    'npxG+xAG.1': 'npxG+xAG_per90'
}, inplace=True)

# Salvando dados na pasta processed
df_dados_matches.to_csv('../data/processed/df_dados_matches.csv', index=False)
df_dados_players.to_csv('../data/processed/df_dados_players.csv', index=False)
df_dados_teams.to_csv('../data/processed/df_dados_teams.csv', index=False) 
```
